# Object Tracking with MATA

Demonstrates frame-by-frame object tracking with persistent track IDs
displayed in the `VisionResult` auto-display table.

**Model used**: `facebook/detr-resnet-50` + BotSort tracker

**Prerequisites**: `pip install datamata[notebook]`

```python
# !pip install datamata[notebook]
```

```python
import mata
print(f"MATA version: {mata.__version__}")
```

```python
# Load a tracking adapter
tracker = mata.load(
    "track",
    "facebook/detr-resnet-50",
    tracker="botsort",
    conf=0.3,
)
```

```python
import cv2

VIDEO = "../../data/assets/sample_video.mp4"  # replace with your video
cap = cv2.VideoCapture(VIDEO)

results = []
for _ in range(10):  # process first 10 frames
    ret, frame = cap.read()
    if not ret:
        break
    frame_rgb = cv2.cvtColor(frame, cv2.COLOR_BGR2RGB)
    result = tracker.update(frame_rgb, persist=True)
    results.append(result)

cap.release()
print(f"Processed {len(results)} frames")
```

```python
# Auto-display the last frame's tracking result (includes Track ID column)
results[-1]
```

- Use `stream=True` for constant-memory streaming on long videos
